**Handle Missing Values**

```python
missing_value = df.isnull().sum()

print(missing_value)

#visualizing missing values

missing_value[missing_value > 0].plot(kind = 'bar',figsize = (8,4))
plt.title("Missing Values")
plt.ylabel("Count")
plt.show()
```

```python
print("Duplicated Rows", df.duplicated().sum())

df.drop_duplicates(inplace = True)

print("New Shape ", df.shape)
```

```python
import ast

#here we have to extracet cpu and memory from f+dictionary strings0
def extract_cpu(value):
    try:
        return ast.literal_eval(value)['cpus']
    except:
        return 0

def extract_memory(value):
    try:
        return ast.literal_eval(value)['memory']
    except:
        return 0

df["avg_cpu"] = df['average_usage'].apply(extract_cpu)
df["avg_memory"] =df["average_usage"].apply(extract_memory)

df["max_cpu"] =df["maximum_usage"].apply(extract_cpu)
df["max_memory"] =df["maximum_usage"].apply(extract_memory)

print(df.head())
df.shape
```

**Distribution of CPU Usage**

```python
df["avg_cpu_percent"] =df["avg_cpu"] * 100

plt.figure(figsize=(7,4))


plt.hist(df["avg_cpu_percent"], bins = 30, edgecolor = "black")
plt.title("Distribution of CPU Utilization")
plt.xlabel("CPU Utilization(%)")
plt.ylabel("Frequency")

plt.xlim(0,100) 
plt.grid(axis="y",alpha=0.3)

plt.show()

```

We conclude that the dataset is skewed to the rightand cloud resourses do not utilize the data more than 20 percent.

**Distribution of CPU Usage Over the day**

```python
df["datetime"] =pd.to_datetime(df["start_time"], unit = "us")

df["hour"] = df["datetime"].dt.hour

hourly_cpu = df.groupby("hour")["avg_cpu"]


hours = hourly_cpu.index.to_numpy()
cpu = hourly_cpu.to_numpy()

plt.figure(figsize = (10,5))

plt.plot(hours, 
         cpu, 
         linewidth = 2, 
         color = "pink",
         marker ="o")

plt.title("Average CPU Utilization Over Hours of the Day")
plt.xlabel("Hour of the Day")
plt.ylabel("Average CPU Utilization %")



plt.xticks(range(24))
plt.ylim(0,2.5)
plt.grid(axis="y", alpha = 0.3) #alpha is used for grid to make it faint

plt.show()
```

The highest utilization is 1.9 % for 16:00 hours

**Distribution of Memory Usage**

```python
plt.figure(figsize=(7,4))

df["memory_usage_percent"] = df["avg_memory"] * 100
plt.hist(df["memory_usage_percent"],bins = 30, edgecolor = "black")
plt.title("Memory Utilization (%)")
plt.xlabel("Memory Usage(%)")
plt.ylabel("Frequency")

plt.show()
```

**Boxplots(Outliner detection)**

```python
plt.figure(figsize=(8,5))

plt.boxplot([df["avg_cpu"], df["avg_memory"]], 
            tick_labels = ["CPU", "Memory"])

plt.title("CPU & Memory Outliners")

plt.show()
```

-The boxplot revels that both CPU AND memory utilization are highly right-skewed. The median value are close to zero, indicating the most cloud consume very few resources. 

-There are many outliners , expecially for CPU Utilization, representing a small number of resource-intensive task.

-These variation justify the need for predictive resource allocation in 
cloud environments.

**Correlation Heatmap**

```python
corr =df[[
    "avg_cpu",
    "avg_memory",
    "max_cpu",
    "max_memory",
    "assigned_memory",
    "failed"
]].corr()

plt.figure(figsize =(7,6))

plt.imshow(corr, cmap="Blues")
plt.colorbar()

labels = corr.columns.tolist()
plt.xticks(range(len(corr.columns)), labels,rotation= 45)
plt.yticks(range(len(corr.columns)), labels)

plt.title("Correlation Heatmap")

plt.show()
```

-The correlation heatmap shows strong positive relationships between average and maximum CPU utilization, as well as between average and maximum memory utilization. 

-Assigned memory is also highly correlated with memory usage, indicating that memory allocation closely reflects resource demand. 

-In contrast, the task failure variable exhibits negligible correlation with the other features, suggesting it contributes little to predicting resource utilization.

**CPU VS Memory Scatter Plot**

```python
plt.figure(figsize=(6,5))

plt.scatter(df["avg_cpu"], df["avg_memory"], alpha = 0.3)

plt.xlabel("Average CPU")
plt.ylabel("Average Memory")

plt.title("CPU vs Memory Usage")

plt.show()
```

-Most tasks are clustered near the bottom-left, meaning they use low CPU and low memory.

-There is a moderate positive relationship: as CPU utilization increases, memory utilization generally tends to increase as well.

-A few points in the upper-right corner are resource-intensive outliers, representing tasks that consume both high CPU and high memory.

-The wide spread of points at low CPU values indicates that tasks with similar CPU usage can still have different memory requirements.

```python
failure = df["failed"].value_counts()

failure.plot(kind="bar", figsize=(5,4))

plt.title("Task Failure Distribution")
plt.xlabel("Failed")
plt.ylabel("Count")

plt.show()

print(failure)
```

-Most tasks were completed successfully (failed = 0).

-A smaller proportion of tasks failed (failed = 1).

-The dataset is imbalanced, with successful tasks greatly outnumbering failed tasks.

-This suggests that task failures are relatively infrequent and may have limited influence on predicting resource utilization.

```python
target_corr = corr["avg_cpu"].sort_values(ascending=False)

print(target_corr)

target_corr.plot(kind="bar", figsize=(6,4))

plt.title("Feature Correlation with CPU Usage")
plt.ylabel("Correlation")

plt.show()
```

-max_cpu has the strongest correlation with avg_cpu (0.92), meaning machines with higher peak CPU usage usually have higher average CPU usage.

-avg_memory and max_memory have moderate positive correlations (0.48), showing that CPU and memory usage tend to increase together.

-assigned_memory has a weak positive correlation (0.31), so memory allocation has only a limited relationship with CPU utilization.

-failed has almost zero correlation, indicating that task failure is not a useful predictor of CPU usage.
